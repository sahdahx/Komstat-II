## **1. STUDI KASUS K-MEDOIDS (PAM)**

Sebuah kedai kopi ingin mengelompokkan pelanggan berdasarkan Pengeluaran per bulan (juta rupiah) dan Frekuensi kunjungan per bulan. Manajemen ingin membagi pelanggan ke dalam 2 kelompok untuk menyusun strategi pemasaran yang berbeda.

Data pelanggan:

| Pelanggan | Pengeluaran (juta Rp) | Frekuensi Kunjungan |
|-----------|----------------------|---------------------|
| A | 1 | 1 |
| B | 2 | 1 |
| C | 2 | 2 |
| D | 8 | 8 |
| E | 9 | 8 |
| F | 8 | 9 |

Gunakan metode K-Medoids (PAM) dengan jarak Euclidean dan jumlah cluster k = 2 untuk mengelompokkan pelanggan.

## Konsep Singkat

K-Medoids merupakan metode partition clustering yang mirip dengan K-Means. Perbedaannya, pusat cluster menggunakan objek data asli (medoid), bukan rata-rata data (centroid). Metode ini lebih tahan terhadap outlier karena pusat cluster selalu merupakan salah satu observasi dalam dataset. Pada studi kasus ini digunakan jarak Euclidean untuk mengukur kedekatan antar pelanggan dan membentuk dua kelompok pelanggan dengan karakteristik yang mirip.


````python
import pandas as pd
import numpy as np
from sklearn_extra.cluster import KMedoids

# Data pelanggan
X = np.array(
    [
        [1,1],
        [2,1],
        [2,2],
        [8,8],
        [9,8],
        [8,9]
    ]
)
data = pd.DataFrame(X, columns=["Pengeluaran","Frekuensi"])
data
````

````python
# Membuat model K-Medoids
kmedoids = KMedoids(
    n_clusters=2,
    metric="euclidean",
    random_state=42
)

# Menyesuaikan model dengan data
kmedoids.fit(X)

# Menampilkan hasil clustering
print(f"\nMedoid:\n{kmedoids.cluster_centers_}")
print(f"\nCluster:\n{kmedoids.labels_}")
````

**Penjelasan:**

Hasil clustering menunjukkan bahwa algoritma K-Medoids membentuk **2 cluster**. Medoid cluster pertama adalah titik **(2,1)** yang merupakan pelanggan **B**, sedangkan medoid cluster kedua adalah titik **(8,8)** yang merupakan pelanggan **D**. Karena medoid merupakan objek asli dalam data, kedua pelanggan tersebut menjadi pusat representatif masing-masing cluster.

Berdasarkan label cluster:

```text
A → Cluster 0
B → Cluster 0
C → Cluster 0
D → Cluster 1
E → Cluster 1
F → Cluster 1
```

Maka diperoleh:

* **Cluster 0:** A, B, C
* **Cluster 1:** D, E, F

Cluster 0 berisi pelanggan dengan pengeluaran dan frekuensi kunjungan yang relatif rendah. Sementara itu, Cluster 1 berisi pelanggan dengan pengeluaran dan frekuensi kunjungan yang tinggi.

**Interpretasi bisnis:**

* Cluster 0 dapat diberikan promo diskon atau cashback untuk meningkatkan aktivitas belanja.
* Cluster 1 dapat diberikan program loyalitas, membership premium, atau penawaran eksklusif karena merupakan pelanggan bernilai tinggi.

## **2. STUDI KASUS HIERARCHICAL CLUSTERING (AGGLOMERATIVE COMPLETE LINK)**

Sebuah perusahaan ingin mengelompokkan cabang tokonya berdasarkan karakteristik operasional. Variabel yang digunakan adalah Penjualan bulanan (X) dan Jumlah pelanggan (Y).

Data cabang toko:

| Cabang | Penjualan | Jumlah Pelanggan |
|---------|-----------|------------------|
| A | 1 | 1 |
| B | 2 | 1 |
| C | 5 | 5 |
| D | 6 | 5 |

Gunakan metode Hierarchical Clustering dengan algoritma Agglomerative dan Complete Link menggunakan jarak Euclidean untuk menentukan kelompok cabang yang memiliki karakteristik serupa.

## Konsep Singkat

Hierarchical Clustering adalah metode clustering yang membentuk struktur bertingkat (hierarki) dalam bentuk dendrogram. Pada algoritma Agglomerative, setiap data awalnya dianggap sebagai cluster tersendiri, kemudian cluster yang paling dekat digabung secara bertahap hingga terbentuk cluster yang lebih besar. Metode Complete Link mengukur jarak antar cluster menggunakan pasangan titik yang memiliki jarak terjauh. Pada studi kasus ini digunakan jarak Euclidean untuk menghitung kedekatan antar cabang toko sebelum proses penggabungan cluster dilakukan.


````python
import pandas as pd
import numpy as np
from sklearn.cluster import AgglomerativeClustering

# Data cabang toko
X = np.array([
    [1,1],
    [2,1],
    [5,5],
    [6,]
])
data = pd.DataFrame(X, columns=["Penjualan","Pelanggan"])
data
````

````python
# Membuat model Hierarchical Clustering
hc = AgglomerativeClustering(
    n_clusters=2,
    linkage="complete"
)
# Membentuk cluster
labels = hc.fit_predict(X)

# Menampilkan hasil clustering
print(f"\nCluster:\n{labels}")
````

**Penjelasan:**

Hasil clustering menunjukkan bahwa algoritma Hierarchical Clustering dengan metode Agglomerative Complete Link membentuk **2 cluster**.

Berdasarkan label cluster:

```text
A = Cluster 1
B = Cluster 1
C = Cluster 0
D = Cluster 0
```

Maka,  diperoleh **Cluster 1** adalah A, B, sedangkan **Cluster 0** adalah C, D.
Hasil tersebut menunjukkan bahwa cabang A dan B memiliki karakteristik yang mirip karena memiliki penjualan dan jumlah pelanggan yang relatif rendah. Sementara cabang C dan D berada dalam kelompok yang sama karena memiliki penjualan dan jumlah pelanggan yang lebih tinggi.

**Interpretasi bisnis:**

* Cluster A-B merepresentasikan cabang dengan aktivitas operasional yang masih rendah sehingga dapat menjadi fokus peningkatan pemasaran.
* Cluster C-D merepresentasikan cabang dengan performa lebih baik sehingga strategi yang dilakukan dapat difokuskan pada mempertahankan pelanggan dan meningkatkan loyalitas pelanggan yang sudah ada.
