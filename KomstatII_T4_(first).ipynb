```python
# IMPORT LIBRARY
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_squared_error, r2_score

# MEMBANGKITKAN DATA
np.random.seed(42)
jumlah_data = 100

X1 = np.random.normal(0, 1, jumlah_data)
X2 = 0.8 * X1 + np.random.normal(0, 0.2, jumlah_data)
X3 = np.random.normal(0, 1, jumlah_data)
X4 = np.random.normal(0, 1, jumlah_data)
X5 = np.random.normal(0, 1, jumlah_data)

# VARIABEL TARGET
y = 3*X1 + 2*X2 + 1.5*X4 + np.random.normal(0, 1, jumlah_data)

# MEMBUAT DATAFRAME
data = pd.DataFrame({
    "X1": X1,
    "X2": X2,
    "X3": X3,
    "X4": X4,
    "X5": X5,
    "y": y
})
print(data.head())
```

```python
# MEMISAHKAN VARIABEL PREDIKTOR DAN RESPON
X = data.drop(columns="y")
y = data["y"]

# SPLIT DATA TRAINING DAN TESTING
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=10
)

# STANDARDISASI DATA
normalisasi = StandardScaler()
X_train_std = normalisasi.fit_transform(X_train)
X_test_std = normalisasi.transform(X_test)
```

```python
# RIDGE REGRESSION
model_ridge = Ridge(alpha=1)
model_ridge.fit(X_train_std, y_train)
pred_ridge = model_ridge.predict(X_test_std)
print("\nHASIL RIDGE REGRESSION")

for nama, nilai in zip(X.columns, model_ridge.coef_):
    print(f"Koefisien {nama} : {nilai:.4f}")

print(f"Intercept        : {model_ridge.intercept_:.4f}")
print(f"MSE              : {mean_squared_error(y_test, pred_ridge):.4f}")
print(f"R2 Score         : {r2_score(y_test, pred_ridge):.4f}")

# LASSO REGRESSION
model_lasso = Lasso(alpha=0.1)
model_lasso.fit(X_train_std, y_train)
pred_lasso = model_lasso.predict(X_test_std)
print("\nHASIL LASSO REGRESSION")

for nama, nilai in zip(X.columns, model_lasso.coef_):
    print(f"Koefisien {nama} : {nilai:.4f}")
print(f"Intercept        : {model_lasso.intercept_:.4f}")
print(f"MSE              : {mean_squared_error(y_test, pred_lasso):.4f}")
print(f"R2 Score         : {r2_score(y_test, pred_lasso):.4f}")

# ELASTIC NET
model_elastic = ElasticNet(alpha=0.1, l1_ratio=0.5)
model_elastic.fit(X_train_std, y_train)
pred_elastic = model_elastic.predict(X_test_std)
print("\nHASIL ELASTIC NET")

for nama, nilai in zip(X.columns, model_elastic.coef_):
    print(f"Koefisien {nama} : {nilai:.4f}")
print(f"Intercept        : {model_elastic.intercept_:.4f}")
print(f"MSE              : {mean_squared_error(y_test, pred_elastic):.4f}")
print(f"R2 Score         : {r2_score(y_test, pred_elastic):.4f}")
```

## Interpretasi:

### Ridge Regression

Metode Ridge memberikan penalti berbasis L2 sehingga koefisien prediktor diperkecil tetapi tidak sampai nol. Variabel X1, X2, dan X4 memiliki koefisien terbesar sehingga menjadi variabel yang paling berpengaruh terhadap variabel target. Nilai MSE yang kecil dan R² yang tinggi menunjukkan bahwa model mampu menjelaskan variasi data dengan baik.

Pada model Ridge, variabel X1 dan X4 memiliki pengaruh positif terbesar terhadap target, sedangkan X3 dan X5 memiliki pengaruh negatif kecil. Nilai R² sebesar 0,9338 menunjukkan bahwa model mampu menjelaskan sekitar 93,38% variasi data, dengan MSE sebesar 1,3864 yang menandakan error model cukup kecil.

### Lasso Regression

Metode Lasso menggunakan penalti L1 yang dapat mengecilkan bahkan menghilangkan koefisien variabel tertentu. Pada hasil model, koefisien X3 dan X5 menjadi sangat kecil mendekati nol sehingga kedua variabel tersebut dianggap kurang berpengaruh terhadap target. Nilai R² yang tinggi menunjukkan performa model tetap baik walaupun melakukan seleksi variabel.

Pada model Lasso, koefisien X3 dan X5 menjadi sangat kecil mendekati nol sehingga kedua variabel dianggap kurang berpengaruh terhadap target. Variabel yang paling dominan tetap X1 dan X4. Nilai R² sebesar 0,9396 menunjukkan performa model yang sangat baik dengan error lebih kecil dibanding Ridge, yaitu MSE sebesar 1,2648.

### Elastic Net

Elastic Net merupakan kombinasi antara Ridge dan Lasso sehingga mampu menangani multikolinearitas sekaligus melakukan seleksi variabel. Hasil koefisien menunjukkan bahwa X1, X2, dan X4 tetap menjadi variabel utama, sedangkan X3 dan X5 memiliki pengaruh kecil. Model ini cocok digunakan ketika terdapat korelasi antarvariabel prediktor.

Pada model Elastic Net, variabel X1, X2, dan X4 tetap menjadi faktor utama yang memengaruhi target. Nilai R² sebesar 0,9409 merupakan yang tertinggi di antara ketiga model, sedangkan MSE sebesar 1,2385 merupakan yang terendah. Hal ini menunjukkan bahwa Elastic Net memberikan performa prediksi terbaik pada data ini.

### Kesimpulan

Ketiga metode regularisasi menghasilkan performa model yang baik karena memiliki nilai R² tinggi dan MSE rendah. Ridge lebih fokus mengurangi overfitting dengan mengecilkan koefisien, Lasso dapat melakukan seleksi variabel, sedangkan Elastic Net menggabungkan kelebihan keduanya. Pada data ini, variabel yang paling memengaruhi target adalah X1, X2, dan X4.
